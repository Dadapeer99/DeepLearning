# Automatic Differentiation

## partial differential with respect to xand with respect to y

```python
x = tf.Variable(3.0)
y = tf.Variable(4.0)

with tf.GradientTape() as tape:
    z = x**2 + y**3

grads = tape.gradient(z, [x, y])
```

```python
grads
```

# Simple neural network

```python
import numpy as np
import tensorflow as tf

x_train = np.array([[1], [2], [3], [4], [5]], dtype=np.float32)
y_train = np.array([[3], [5], [7], [9], [11]], dtype=np.float32)
```

#### Sequential API is a Keras model type that builds a neural network as a linear stack of layers, 
#### where each layer has exactly 

#### one input and one output,

```python
model = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dense(1)
])
```

```python
# model.compile(
#     optimizer="adam",
#     loss="mse",
#     metrics=["mae"]
# )
model.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
    loss="mse"
)
```

```python
model.fit(x_train, y_train, epochs=10)
```

```python
model.predict(np.array([[6]], dtype=np.float32))
```

# Functional API
multiple inputs->multiple outputs

```python
import numpy as np
import tensorflow as tf

x_train = np.array([[1], [2], [3], [4], [5]], dtype=np.float32)
y_train = np.array([[3], [5], [7], [9], [11]], dtype=np.float32)
inputs = tf.keras.Input(shape=(1,))
outputs = tf.keras.layers.Dense(1)(inputs)

#Functional
model = tf.keras.Model(inputs=inputs, outputs=outputs)

model.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
    loss="mse"
)

model.fit(x_train, y_train, epochs=500, verbose=0)

print(model.predict(np.array([[6]], dtype=np.float32)))
```

# Softmax

```python
import tensorflow as tf

# Raw output from the model (Logits)
# Dog: 2.0, Cat: 1.0, Bird: 0.1
logits = tf.constant([2.0, 1.0, 0.1])

probabilities = tf.nn.softmax(logits)

print(probabilities.numpy())
```

```python
prediction = tf.argmax(probabilities)

print(prediction.numpy())
#If the output is 0, the final prediction is Dog.
#If the output is 1, it’s Cat.
#If the output is 2, it’s Bird.
```
